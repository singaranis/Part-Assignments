```python
import numpy as np
```

```python
#Create a 1D NumPy array named temperatures_w1 for Week 1 with the following
#values:[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# Inspect the shape, data type, and the number of elements in the array. Inspect
#the shape of the temperatures_w1 array.

print("Shape:", temperatures_w1.shape)
print("Datatype:", temperatures_w1.dtype)
print("No of elements:", temperatures_w1.size)
```

```python
#Convert the temperatures_w1 from Celsius to Fahrenheit using the formula:
#Fahrenheit = (Celsius * 9/5) + 32

temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:")
print(temperatures_f)

#Find the maximum, minimum, and mean temperatures for the week.
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
#Extract the temperatures for the first three days of the week.
#Extract the temperatures for the weekend (last two days).
#Extract the temperatures for the middle three days of the week.
print("First three days:", temperatures_w1[:3])
print("Weekend temperatures:", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

```python
#Create a 2D NumPy array named temperatures where each row represents temperatures for a different week:
#★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
#★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])
print("Two-Week Temperatures:")
print(temperatures)
```

```python
#Inspect and print the shape, data type, and total number of elements in the temperatures array.
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

#Extract the temperatures for each week and the weekends (last two days) for both weeks.
print("Week 1:", temperatures[0])
print("Week 2:", temperatures[1])

print("Week 1 Weekend:", temperatures[0, -2:])
print("Week 2 Weekend:", temperatures[1, -2:])
```

```python
#Create a Pandas Series named marks with the following data:
# Marks: 95, 92, 89, 85, 80
#Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
import pandas as pd
marks = pd.Series( [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print(marks)
```

```python
#Use the integer index position to access the mark of the 1st rank student.
print("First Rank Mark:", marks.iloc[0])
#Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
print("Top 3 Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
#Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
print("Third Rank Mark:", marks.iloc[2])
#Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
#Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1'] = 100
print("After updating rank1:")
print(marks)

#Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')
print("After Removing Rank5:")
print(marks)

```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python
#Create a Pandas DataFrame named transactions with the following data: ➢
#TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
#ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
#'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
#➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South','North'
#➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
#● Display the transactions DataFrame and its basic information, including head, tail,
#shape, column names, and data types.
print(transactions)
print(transactions.head())
print(transactions.tail())
print(transactions.shape)
print(transactions.columns)
print(transactions.dtypes)
transactions.info()

#● Display only the 'ProductCategory' and 'Amount' columns.
print(transactions[['ProductCategory', 'Amount']])

#● Retrieve the last 3 columns of the DataFrame using iloc or loc.
print(transactions.iloc[:, -3:])

#● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)

#● Find the value counts for the 'ProductCategory' column.
print(transactions['ProductCategory'].value_counts())

#● Find the unique values in the 'Region' column.
print(transactions['Region'].unique())

#● Group by 'Region' and find the mean amount for each region.
mean_amount = transactions.groupby('Region')['Amount'].mean()
print(mean_amount)
```

```python
#● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)

#● Add a new column 'Discount' by calculating 10% of the 'Amount'.
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)

#● Remove the row with TransactionID 109.
transactions = transactions[
    transactions['TransactionID'] != 109
].reset_index(drop=True)

print(transactions)

#● Delete the 'Discount' column from the DataFrame.
transactions = transactions.drop(columns=['Discount'])

print("Final Transactions DataFrame:")
print(transactions)
```
